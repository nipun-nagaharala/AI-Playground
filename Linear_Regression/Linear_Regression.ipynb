```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
```

```python
df = pd.read_csv('w2_position_salaries.csv')

print(df.head())
print(df.columns)
print(df.shape)
```

Linear Regression

```python
from sklearn import linear_model
#Separate the input from the output

X=df[['Level']]
y=df['Salary']


linear_model = LinearRegression()
linear_model.fit(X,y)
y_pred=linear_model.predict(X)
r2_linear=linear_model.score(X,y)

print("R2 Score :",r2_linear)
print("Y Prediction :",y_pred)
```

Polynormal Regression

```python
#Transform the input features
poly2 = PolynomialFeatures(degree=2)
X_poly2=poly2.fit_transform(X)

#Train the polynomial model
model2 = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)

model2.fit(X,y)

r2_degree=model2.score(X,y)

print("R2 Degree :",r2_degree)

```

```python
#For degrees 3 and 4

for degree in [2, 3, 4]:
    model = make_pipeline(
        PolynomialFeatures(
            degree=degree,
            include_bias=False
        ),
        LinearRegression()
    )

    model.fit(X, y)

    r2 = model.score(X, y)

    print("Degree:", degree, "R2:", r2)
```
